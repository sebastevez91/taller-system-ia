```python
import pandas as pd

# Simulamos algunos mantenimientos, como saldrían de tu tabla Mantenimiento
mantenimientos = pd.DataFrame({
    "vehiculo": ["Corolla", "Civic", "Corolla", "Civic", "Corolla"],
    "componente": ["bateria", "correa", "correa", "bateria", "pastillas"],
    "km_al_momento": [45000, 60000, 90000, 30000, 38000],
    "costo": [45000.50, 80000, 95000, 42000, 20000],
})
mantenimientos
```

```python
# Todos los mantenimientos del Corolla
mantenimientos[mantenimientos["vehiculo"] == "Corolla"]
```

```python
# Costo total gastado por componente, a través de todos los vehículos
mantenimientos.groupby("componente")["costo"].sum()
```

```python
# Ejemplo: "porcentaje de vida útil consumido" si la correa dura 90000km
mantenimientos["pct_vida_util"] = mantenimientos["km_al_momento"] / 90000
mantenimientos
```

```python
# Filtro solo los mantenimientos con costo mayor a 40000
mantenimientos[mantenimientos["costo"] > 40000]
```

```python
filtrado = mantenimientos[mantenimientos["costo"] > 40000]
filtrado.groupby("vehiculo")["costo"].mean()
```

```python
import numpy as np
import pandas as pd

np.random.seed(42) # Para que los resultados sean reproducibles entre corridas

componentes = pd.DataFrame({
    "componente": ["bateria", "correa_distribucion", "pastillas_freno", "amortiguadores", "filtro_aceite"],
    "vida_util_km": [60000, 90000, 30000, 80000, 10000],
    "vida_util_meses": [36, 60, 24, 48, 12],
})

N = 2000 # Cantidad de registros a simular

# Elegimos un componente al azar para cada registro
idx_componente = np.random.randint(0, len(componentes), N)
comp_asignado = componentes.iloc[idx_componente].reset_index(drop=True)

# Simulamos cuanto km lleva ese componente desde su instalacion (entre 0 y 1.3x su vida util)
km_desde_instalacion = np.random.uniform(0, comp_asignado["vida_util_km"] * 1.3)

dataset = pd.DataFrame({
    "componente": comp_asignado["componente"],
    "vida_util_km": comp_asignado["vida_util_km"],
    "km_desde_instalacion": km_desde_instalacion.round(0),
})

dataset["pct_vida_util"] = dataset["km_desde_instalacion"] / dataset["vida_util_km"]
dataset.head(10)
```

```python
import numpy as np
import pandas as pd

np.random.seed(42) # Para que los resultados sean reproducibles entre corridas

componentes = pd.DataFrame({
    "componente": ["bateria", "correa_distribucion", "pastillas_freno", "amortiguadores", "filtro_aceite"],
    "vida_util_km": [60000, 90000, 30000, 80000, 10000],
    "vida_util_meses": [36, 60, 24, 48, 12],
})

N = 2000 # Cantidad de registros a simular

# Elegimos un componente al azar para cada registro
idx_componente = np.random.randint(0, len(componentes), N)
comp_asignado = componentes.iloc[idx_componente].reset_index(drop=True)

# Simulamos cuanto km lleva ese componente desde su instalacion (entre 0 y 1.3x su vida util)
km_desde_instalacion = np.random.uniform(0, comp_asignado["vida_util_km"] * 1.3)

dataset = pd.DataFrame({
    "componente": comp_asignado["componente"],
    "vida_util_km": comp_asignado["vida_util_km"],
    "km_desde_instalacion": km_desde_instalacion.round(0),
})

# Probabilidad de falla: crece con pct_vida_util, con curva suave
# En pct=0 la probabilidad es casi 0, en pct=1 (100% de vida util) ronda 50%, en pct=1.3 supera 80%
prob_falla = 1 / (1 + np.exp(-6 * (dataset["pct_vida_util"] - 1)))

# "Tiramos la moneda" cargada con esa probabilidad para cada fila
dataset["fallo"] = (np.random.uniform(0, 1, len(dataset)) < prob_falla).astype(int)

dataset[["componente", "pct_vida_util", "fallo"]].sort_values("pct_vida_util").head(15)
```
